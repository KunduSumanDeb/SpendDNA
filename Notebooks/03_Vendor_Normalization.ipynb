# Vendor Normalization

This notebook standardizes merchant/vendor names from transaction descriptions.

**Input**

- cleaned_data.pkl

**Output**

- vendor_data.pkl

```python
# Setup

import os
import sys

PROJECT_ROOT = os.path.abspath("..")

if PROJECT_ROOT not in sys.path:
    sys.path.append(PROJECT_ROOT)
```

```python
# Imports

import pandas as pd

from Config.paths import CLEANED_DATA_PKL, VENDOR_DATA_PKL

from Utils.io import load_pickle, save_pickle
from Utils.validation import validate_dataframe
from Utils.vendor import normalize_vendors
```

```python
# Load Dataset

transactions = load_pickle(CLEANED_DATA_PKL)

print(f"Rows Loaded    : {len(transactions)}")
print(f"Columns Loaded : {transactions.shape[1]}")

transactions.head()
```

```python
# Normalize Vendor Names

vendor_data = normalize_vendors(transactions)

vendor_data.head(10)
```

# Validation

```python
# Validate Dataset

validate_dataframe(vendor_data)
```

```python
# Vendor Validation

print("=" * 50)
print("Vendor Normalization Summary")
print("=" * 50)

print(f"Total Transactions : {len(vendor_data)}")
print(f"Unique Vendors     : {vendor_data['Vendor'].nunique()}")

print("\nTop 25 Vendors")
print("-" * 50)

print(vendor_data["Vendor"].value_counts().head(25))
```

```python
save_pickle(vendor_data, VENDOR_DATA_PKL)

print("Vendor dataset exported successfully.")
```
