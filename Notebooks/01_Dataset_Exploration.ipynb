```python
# Import required libraries

from pathlib import Path

import pandas as pd
import numpy as np
```

```python
# Configure project directories

PROJECT_ROOT = Path.cwd().parent

DATASET_DIR = PROJECT_ROOT / "Dataset"
DATA_DIR = PROJECT_ROOT / "Data"
OUTPUT_DIR = PROJECT_ROOT / "Outputs"
CONFIG_DIR = PROJECT_ROOT / "Config"

RAW_DATASET = DATASET_DIR / "raw" / "Dataset_Raw.csv"
RAW_DATA_PKL = DATA_DIR / "raw_data.pkl"
```

```python
# Verify important project paths

print(f"Project Root : {PROJECT_ROOT}")
print(f"Dataset Path : {RAW_DATASET}")
print(f"Dataset Exists : {RAW_DATASET.exists()}")
```

```python
# Load the transaction dataset

def load_dataset(path: Path) -> pd.DataFrame:
    """
    Load the raw transaction dataset.

    Parameters
    ----------
    path : Path
        Path to the CSV file.

    Returns
    -------
    pd.DataFrame
        Loaded transaction dataset.
    """

    return pd.read_csv(path)
```

```python
# Read the dataset

df = load_dataset(RAW_DATASET)

print("Dataset loaded successfully.\n")
print(f"Rows    : {df.shape[0]}")
print(f"Columns : {df.shape[1]}")
```

```python
# Display the first five records

display(df.head())
```

## Explore Dataset Structure

In this section, we examine the overall structure of the dataset, including
its dimensions, column names, data types, memory usage, and basic statistics.
This helps us understand the dataset before performing any preprocessing.

```python
# Display dataset dimensions

print("Dataset Shape")
print("-" * 40)
print(f"Rows    : {df.shape[0]}")
print(f"Columns : {df.shape[1]}")
```

```python
# Display column names

print("Dataset Columns")
print("-" * 40)

for index, column in enumerate(df.columns, start=1):
    print(f"{index}. {column}")
```

```python
# Display dataset information

df.info()
```

```python
# Display statistical summary

display(df.describe(include="all"))
```

## Data Quality Analysis

This section evaluates the quality of the dataset by identifying missing values,
duplicate records, duplicate transaction references, and the uniqueness of each
column. These observations will guide the data cleaning process in the next
notebook.

```python
# Check missing values

missing_values = df.isnull().sum().to_frame(name="Missing Values")
missing_values["Percentage"] = (
    missing_values["Missing Values"] / len(df) * 100
).round(2)

display(missing_values)
```

```python
# Check duplicate rows

duplicate_rows = df.duplicated().sum()

print("Duplicate Rows")
print("-" * 40)
print(f"Total Duplicate Rows : {duplicate_rows}")
```

```python
# Check duplicate transaction references

duplicate_refs = df["Ref"].duplicated().sum()

print("Duplicate Transaction References")
print("-" * 40)
print(f"Duplicate References : {duplicate_refs}")
```

```python
# Display unique values in each column

unique_values = df.nunique().to_frame(name="Unique Values")

display(unique_values)
```

## Key Findings

Based on the exploratory analysis, the following observations were made:

- The dataset contains transaction records with eight attributes.
- Multiple date formats are present within the `Date` column.
- The `Amount` column contains currency symbols and formatting, preventing it from being interpreted as a numeric field.
- The `Type` column contains inconsistent labels (such as `DR`, `CR`, and `Debit`).
- The `Balance` column is already stored as a numeric data type.
- Transaction descriptions appear in multiple formats and will require vendor normalization.
- Duplicate transaction references, if any, should be investigated during the data cleaning stage.
- These issues will be addressed in **Notebook 02: Data Cleaning**.

```python
# Validate dataset before export

def validate_dataset(dataframe: pd.DataFrame) -> None:
    """
    Validate the loaded dataset before exporting.
    """

    print("Validation Summary")
    print("-" * 40)

    print(f"Rows                 : {dataframe.shape[0]}")
    print(f"Columns              : {dataframe.shape[1]}")
    print(f"Missing Values       : {dataframe.isnull().sum().sum()}")
    print(f"Duplicate Rows       : {dataframe.duplicated().sum()}")
    print(f"Duplicate References : {dataframe['Ref'].duplicated().sum()}")

    print("\nValidation Completed Successfully.")
```

```python
# Execute validation

validate_dataset(df)
```
