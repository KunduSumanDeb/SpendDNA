# Function 8
## Spending Archetype Detection

This notebook identifies user spending archetypes based on spending behaviour and category-wise transaction analysis.

```python
# Project paths
from pathlib import Path

PROJECT_DIR = Path.cwd().parent

DATA_DIR = PROJECT_DIR / "Data"

UTILS_DIR = PROJECT_DIR / "Utils"
```

```python
# Import libraries

import sys
import pickle
import importlib

import pandas as pd

sys.path.append(str(UTILS_DIR))

import archetype

importlib.reload(archetype)
```

```python
# Load anomaly detection output

with open(
    DATA_DIR / "anomaly_detection.pkl",
    "rb"
) as file:

    anomaly_data = pickle.load(file)

print(anomaly_data.keys()) 
```

```python
# Extract transactions

transactions = anomaly_data["transactions"]

print(
    "Transaction Shape :",
    transactions.shape
)

display(transactions.head())
```

```python
# Generate spending archetypes

archetype_results = (
    archetype.generate_spending_archetypes(
        transactions
    )
)

print(archetype_results.keys())
```

```python
# Display all archetypes

archetypes = pd.DataFrame(
    archetype_results["archetypes"]
)

display(archetypes)
```

```python
# Display matched archetypes

matched = pd.DataFrame(
    archetype_results["matched_archetypes"]
)

display(matched)
```

```python
# Display metadata

metadata = archetype_results["metadata"]

print(metadata)
```

```python
# Validate outputs

assert "transactions" in archetype_results
assert "archetypes" in archetype_results
assert "matched_archetypes" in archetype_results
assert "metadata" in archetype_results

print(
    "Validation completed successfully."
)
```

```python
# Export spending archetypes

with open(
    DATA_DIR / "spending_archetypes.pkl",
    "wb"
) as file:

    pickle.dump(
        archetype_results,
        file
    )

print(
    "Spending archetypes exported successfully."
)
```

```python
# Verify export

with open(
    DATA_DIR / "spending_archetypes.pkl",
    "rb"
) as file:

    exported = pickle.load(file)

print(exported.keys())
```
