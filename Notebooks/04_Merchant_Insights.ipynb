# Function 4: Merchant Insights

This notebook generates merchant-level spending insights from the vendor-normalized transaction dataset.

The output of this notebook is exported as `merchant_summary.pkl`.

```python
# Setup

import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent

if str(PROJECT_ROOT) not in sys.path:
    sys.path.append(str(PROJECT_ROOT))
```

```python
# Imports

import pickle

from Config.paths import (
    VENDOR_DATA_PKL,
    MERCHANT_SUMMARY_PKL,
)

from Utils.merchant import generate_merchant_insights
```

```python
# Load dataset

with open(VENDOR_DATA_PKL, "rb") as file:
    vendor_data = pickle.load(file)

print("Vendor dataset loaded successfully.")
print(f"Rows    : {len(vendor_data)}")
print(f"Columns : {len(vendor_data.columns)}")

print("\nData Types:")
print(vendor_data.dtypes)

print("\nColumns:")
print(vendor_data.columns.tolist())

print("\nFirst 5 Rows:")
print(vendor_data.head())

vendor_data.head()
```

```python
# Generate merchant insights

merchant_results = generate_merchant_insights(vendor_data)

merchant_summary = merchant_results["merchant_summary"]
merchant_category = merchant_results["merchant_category"]
monthly_summary = merchant_results["monthly_summary"]
top_spend = merchant_results["top_spend"]
top_frequency = merchant_results["top_frequency"]
metadata = merchant_results["metadata"]

print("Merchant insights generated successfully.")
```

```python
print(vendor_data.columns.tolist())
```

```python
# Export merchant insights

merchant_summary_data = {
    "merchant_summary": merchant_summary,
    "merchant_category": merchant_category,
    "monthly_summary": monthly_summary,
    "top_spend": top_spend,
    "top_frequency": top_frequency,
    "metadata": metadata,
}

with open(MERCHANT_SUMMARY_PKL, "wb") as file:
    pickle.dump(merchant_summary_data, file)

print("Merchant summary exported successfully.")
```
