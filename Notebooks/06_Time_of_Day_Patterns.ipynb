# Function 6: Time-of-Day Spending Pattern Analysis

## Objective

This notebook analyzes transaction behaviour across different hours of the day.

### Goals

- Extract transaction hour
- Build hourly spending summary
- Create category vs hour spending matrix
- Identify peak spending hour
- Identify peak spending hour for each category
- Analyse late-night spending
- Generate metadata
- Export results for downstream functions

### Input

Data/monthly_trends.pkl

### Output

Data/time_of_day_patterns.pkl

```python
# Project folders and file paths

from pathlib import Path

PROJECT_ROOT = Path.cwd().parent

DATA_DIR = PROJECT_ROOT / "Data"
UTILS_DIR = PROJECT_ROOT / "Utils"

INPUT_FILE = DATA_DIR / "monthly_trends.pkl"
OUTPUT_FILE = DATA_DIR / "time_of_day_patterns.pkl"
```

```python
# Import required libraries

import sys
import pickle
import pandas as pd
import numpy as np

sys.path.append(str(UTILS_DIR))

import time_of_day
```

```python
# Load data from previous function

with open(INPUT_FILE, "rb") as file:
    monthly_data = pickle.load(file)
print("Monthly trends loaded successfully.")
```

```python
# Extract transaction data

transactions = monthly_data["transactions"]
print("Transaction Shape :", transactions.shape)
display(transactions.head())
```

```python
# Run time-of-day analysis

results = time_of_day.generate_time_of_day_patterns(
    transactions
)
print("Time-of-day analysis completed successfully.")
```

```python
# Extract all outputs

hourly_summary = results["hourly_summary"]
category_hour_matrix = results["category_hour_matrix"]
category_peak_hours = results["category_peak_hours"]
late_night_summary = results["late_night_summary"]
peak_hour = results["peak_hour"]
metadata = results["metadata"]
```

```python
# Display analysis results

print("=" * 70)
print("HOURLY SPENDING SUMMARY")
print("=" * 70)

display(hourly_summary)

print()

print("=" * 70)
print("CATEGORY VS HOUR MATRIX")
print("=" * 70)

display(category_hour_matrix)

print()

print("=" * 70)
print("CATEGORY PEAK HOURS")
print("=" * 70)

display(category_peak_hours)

print()

print("=" * 70)
print("OVERALL PEAK HOUR")
print("=" * 70)

print(peak_hour)

print()

print("=" * 70)
print("LATE NIGHT SUMMARY")
print("=" * 70)

print(late_night_summary)
```

```python
# Display hourly spending chart

print("=" * 70)
print("HOURLY SPENDING VISUALIZATION")
print("=" * 70)

time_of_day.print_hourly_bars(
    hourly_summary
)
```

```python
# Validate generated outputs

assert len(hourly_summary) == 24
assert category_hour_matrix.shape[1] == 24
assert "Peak Hour" in peak_hour
assert metadata["total_hours"] == 24
print("Validation completed successfully.")
```

```python
# Save results for next function

time_of_day_data = {
    "transactions": transactions,
    "hourly_summary": hourly_summary,
    "category_hour_matrix": category_hour_matrix,
    "category_peak_hours": category_peak_hours,
    "late_night_summary": late_night_summary,
    "peak_hour": peak_hour,
    "metadata": metadata
}

with open(OUTPUT_FILE, "wb") as file:
    pickle.dump(time_of_day_data, file)

print("Time-of-day patterns exported successfully.")
```

```python
# Print final summary

print("=" * 70)
print("FUNCTION 6 SUMMARY")
print("=" * 70)

print(f"Transactions          : {metadata['total_transactions']}")
print(f"Categories            : {metadata['total_categories']}")
print(f"Hours Analysed        : {metadata['total_hours']}")

print()
print("Peak Spending Hour")
print(peak_hour)
print()
print("Late Night Summary")
print(late_night_summary)
print()
print("Function 6 completed successfully.")
```

```python
# Validate outputs

print("=" * 70)
print("VALIDATION")
print("=" * 70)

assert len(hourly_summary) == 24
assert category_hour_matrix.shape[1] == 24
assert metadata["total_hours"] == 24

print("Rows in Hourly Summary :", len(hourly_summary))
print("Hours Covered          :", metadata["total_hours"])
print("Peak Hour              :", peak_hour["Peak Hour"])
print("Validation completed successfully.")
```
