# Function 9 - Final SpendDNA Report

This notebook generates the consolidated SpendDNA report by combining the outputs of Functions 4–8.

Output:
- final_report.pkl

```python
# Setup
from pathlib import Path

PROJECT_DIR = Path.cwd().parent

DATA_DIR = PROJECT_DIR / "Data"

UTILS_DIR = PROJECT_DIR / "Utils"
```

```python
# Import
import sys
import pickle
import importlib

import pandas as pd

sys.path.append(str(UTILS_DIR))

import report

importlib.reload(report)
```

```python
with open(DATA_DIR / "merchant_summary.pkl", "rb") as file:
    merchant_data = pickle.load(file)

with open(DATA_DIR / "monthly_trends.pkl", "rb") as file:
    monthly_data = pickle.load(file)

with open(DATA_DIR / "time_of_day_patterns.pkl", "rb") as file:
    time_data = pickle.load(file)

with open(DATA_DIR / "anomaly_detection.pkl", "rb") as file:
    anomaly_data = pickle.load(file)

with open(DATA_DIR / "spending_archetypes.pkl", "rb") as file:
    archetype_data = pickle.load(file)

print("All data loaded successfully.")
```

```python
final_report = report.build_report(
    merchant_data,
    monthly_data,
    time_data,
    anomaly_data,
    archetype_data
)

print(final_report.keys())
```

```python
print("EXECUTIVE SUMMARY")
print(final_report["executive_summary"])

print("\nTOP SPENDING VENDORS")
display(final_report["top_spending_vendors"])

print("\nTOP FREQUENCY VENDORS")
display(final_report["top_frequency_vendors"])

print("\nMONTHLY SUMMARY")
display(final_report["monthly_summary"])

print("\nTIME SUMMARY")
print(final_report["time_summary"])

print("\nTOP ANOMALIES")
display(final_report["anomaly_summary"])

print("\nSPENDING ARCHETYPES")
display(final_report["archetype_summary"])
```

```python
assert "executive_summary" in final_report
assert "top_spending_vendors" in final_report
assert "top_frequency_vendors" in final_report
assert "monthly_summary" in final_report
assert "time_summary" in final_report
assert "anomaly_summary" in final_report
assert "archetype_summary" in final_report
assert "metadata" in final_report

print("Validation completed successfully.")
```

```python
with open(
    DATA_DIR / "final_report.pkl",
    "wb"
) as file:

    pickle.dump(
        final_report,
        file
    )

print("final_report.pkl exported successfully.")
```
