# Function 5 : Monthly Trend Analysis

This notebook generates monthly spending trends from the normalized vendor dataset.

### Objectives

- Load the normalized transaction dataset.
- Generate monthly spending trends by category.
- Create a Category × Month spending matrix.
- Identify categories with the highest growth.
- Identify categories with the highest decline.
- Validate all generated outputs.
- Export the monthly trends for downstream analysis.

**Input**

- vendor_data.pkl

**Output**

- monthly_trends.pkl

```python
# Setup paths
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent
DATA_DIR = PROJECT_ROOT / "Data"
UTILS_DIR = PROJECT_ROOT / "Utils"

VENDOR_DATA_PKL = DATA_DIR / "vendor_data.pkl"
MONTHLY_TRENDS_PKL = DATA_DIR / "monthly_trends.pkl"
```

```python
# Import libraries
import sys
import pickle
import numpy as np
import pandas as pd

sys.path.append(str(UTILS_DIR))

from monthly_trends import generate_monthly_trends
```

```python
# Load vendor data
with open(VENDOR_DATA_PKL, "rb") as file:
    vendor_data = pickle.load(file)

print("Vendor data loaded successfully.")
print(f"Rows    : {len(vendor_data)}")
print(f"Columns : {len(vendor_data.columns)}")

vendor_data.head()
```

```python
# Generate monthly trends
monthly_results = generate_monthly_trends(vendor_data)

print("Monthly trends generated successfully.")

for key in monthly_results:
    print("-", key)
```

```python
# Extract outputs
monthly_pivot = monthly_results["monthly_pivot"]
monthly_totals = monthly_results["monthly_totals"]
monthly_changes = monthly_results["monthly_changes"]
growth_summary = monthly_results["growth_summary"]
decline_summary = monthly_results["decline_summary"]
metadata = monthly_results["metadata"]
```

```python
print("Monthly Pivot")
display(monthly_pivot.head())

print("\nMonthly Totals")
display(monthly_totals)

print("\nGrowth Summary")
display(growth_summary.head())

print("\nDecline Summary")
display(decline_summary.head())

print("\nMetadata")
print(metadata)
```

```python
# Validate outputs

print("Validation Summary")
print("-" * 40)

print(f"Transactions       : {metadata['total_transactions']}")
print(f"Categories         : {metadata['total_categories']}")
print(f"Months             : {metadata['total_months']}")

assert not monthly_pivot.empty, "Monthly pivot is empty."
assert not monthly_totals.empty, "Monthly totals are empty."
assert not growth_summary.empty, "Growth summary is empty."
assert not decline_summary.empty, "Decline summary is empty."

print("\nValidation completed successfully.")
```

```python
# Export results

monthly_trends_data = {
    "monthly_pivot": monthly_pivot,
    "monthly_totals": monthly_totals,
    "monthly_changes": monthly_changes,
    "growth_summary": growth_summary,
    "decline_summary": decline_summary,
    "metadata": metadata
}

with open(MONTHLY_TRENDS_PKL, "wb") as file:
    pickle.dump(monthly_trends_data, file)

print("Monthly trends exported successfully.")
```
