# Function 7 : Anomaly Detection

## Objective

Detect anomalous spending transactions using the Z-score method.

### Input

- time_of_day_patterns.pkl

### Output

- anomaly_detection.pkl

### Tasks

- Load previous output
- Detect anomalies
- Validate results
- Export output

```python
# Setup paths

from pathlib import Path

PROJECT_ROOT = Path.cwd().parent

DATA_DIR = PROJECT_ROOT / "Data"
UTILS_DIR = PROJECT_ROOT / "Utils"

INPUT_FILE = DATA_DIR / "time_of_day_patterns.pkl"
OUTPUT_FILE = DATA_DIR / "anomaly_detection.pkl"
```

```python
# Import libraries

import sys
import pickle
import importlib
import pandas as pd
import numpy as np

sys.path.append(str(UTILS_DIR))

import anomaly_detection

importlib.reload(anomaly_detection)
```

```python
# Load previous output

with open(INPUT_FILE, "rb") as file:
    time_data = pickle.load(file)

print("Data loaded successfully.")
print(type(time_data))
```

```python
# Extract transaction data

transactions = time_data["transactions"]

print("Transaction Shape :", transactions.shape)

display(transactions.head())
```

```python
# Run anomaly detection

results = anomaly_detection.generate_anomaly_detection(transactions)

print("Anomaly detection completed successfully.")
```

```python
# Extract outputs

category_statistics = results["category_statistics"]
transactions_with_zscore = results["transactions_with_zscore"]
anomalies = results["anomalies"]
top_anomalies = results["top_anomalies"]
metadata = results["metadata"]
```

```python
# Display results

print("Category Statistics")
display(category_statistics)

print("\nTop Anomalies")
display(top_anomalies)

print("\nMetadata")
print(metadata)
```

```python
# Validate outputs

assert isinstance(category_statistics, pd.DataFrame)
assert isinstance(transactions_with_zscore, pd.DataFrame)
assert isinstance(anomalies, pd.DataFrame)
assert isinstance(top_anomalies, pd.DataFrame)
assert isinstance(metadata, dict)

print("Validation completed successfully.")
```

```python
# Export results

output = {
    "transactions": transactions,
    "category_statistics": category_statistics,
    "transactions_with_zscore": transactions_with_zscore,
    "anomalies": anomalies,
    "top_anomalies": top_anomalies,
    "metadata": metadata
}

with open(OUTPUT_FILE, "wb") as file:
    pickle.dump(output, file)

print("Anomaly detection exported successfully.")
```
